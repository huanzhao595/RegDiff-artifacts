Tool for openai_openai_clasues_pp_maaping_matrix.ipynb
Traverse folders starting with AA under 1122apk\1122apk_privacy_policy_url, then traverse their first-level subfolders and files. Find JSON files ending with urls. The filename contains the apkname followed by the version number. For example, in ae.brandsforless.android-412_urls.json, ae.brandsforless.android is the apkname and 412 is the version. The goal is to store all version numbers for each apkname. The results can be saved as JSON or CSV, and an apkname may have two or more versions.


```python
import os
import re
import json
import csv
from collections import defaultdict

# Root directory
root_dir = r"1122apk\1122apk_privacy_policy_url"

# Output files
json_output = os.path.join(root_dir, "apk_versions_summary.json")
csv_output = os.path.join(root_dir, "apk_versions_summary.csv")

# Filename pattern: apkname-version_urls.json
# Example: ae.brandsforless.android-412_urls.json
pattern = re.compile(r"^(?P<apkname>.+)-(?P<version>\d+)_urls\.json$", re.IGNORECASE)

# Store results: apkname -> set(versions)
apk_versions = defaultdict(set)

# Traverse folders starting with AA under the root directory
for aa_folder in os.listdir(root_dir):
    aa_folder_path = os.path.join(root_dir, aa_folder)

    if not os.path.isdir(aa_folder_path):
        continue
    if not aa_folder.startswith("AA"):
        continue

    # Traverse first-level subfolders under each AA folder
    for subfolder in os.listdir(aa_folder_path):
        subfolder_path = os.path.join(aa_folder_path, subfolder)

        if not os.path.isdir(subfolder_path):
            continue

        # Traverse files in each first-level subfolder
        for filename in os.listdir(subfolder_path):
            if not filename.lower().endswith("_urls.json"):
                continue

            match = pattern.match(filename)
            if not match:
                print(f"Skipping unparseable filename: {os.path.join(subfolder_path, filename)}")
                continue

            apkname = match.group("apkname")
            version = match.group("version")

            apk_versions[apkname].add(version)

# Prepare JSON results
json_data = {
    apkname: sorted(list(versions), key=lambda x: int(x))
    for apkname, versions in sorted(apk_versions.items())
}

with open(json_output, "w", encoding="utf-8") as f:
    json.dump(json_data, f, ensure_ascii=False, indent=2)

# Prepare CSV results
with open(csv_output, "w", encoding="utf-8-sig", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["apkname", "version_count", "versions"])

    for apkname, versions in sorted(apk_versions.items()):
        version_list = sorted(list(versions), key=lambda x: int(x))
        writer.writerow([
            apkname,
            len(version_list),
            ",".join(version_list)
        ])

print(f"Completed; found {len(apk_versions)} apkname entries")
print(f"JSON saved to: {json_output}")
print(f"CSV saved to: {csv_output}")
```

```python
from pathlib import Path
import json

seg_dir = Path(r"1122apk\1122apk_privacy_policy_url\pp_md_2024_10\AA1_first_batch\AA2_second_100_batch_seg_json")

all_segment_texts = []      # Flat list containing only text values
segment_rows = []           # Detailed records with filename/segment_id (optional)
per_file_segments = {}      # List of text values for each file

for fp in sorted(seg_dir.glob("*.json")):
    try:
        data = json.loads(fp.read_text(encoding="utf-8"))
    except Exception as e:
        print(f"[Skip] Failed to read: {fp.name} -> {e}")
        continue

    segments = data.get("segments", [])
    texts = []

    for seg in segments:
        if not isinstance(seg, dict):
            continue
        txt = seg.get("text", "")
        if txt is None:
            txt = ""
        txt = str(txt).strip()
        if not txt:
            continue

        texts.append(txt)
        all_segment_texts.append(txt)
        segment_rows.append({
            "file": fp.name,
            "segment_id": seg.get("segment_id"),
            "text": txt
        })

    per_file_segments[fp.name] = texts

print("Number of JSON files:", len(per_file_segments))
print("Total number of segments.text entries:", len(all_segment_texts))
print("First 3 examples:", all_segment_texts[:3])
```
